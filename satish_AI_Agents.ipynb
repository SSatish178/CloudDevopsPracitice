```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
print(llm.invoke("Hello! Nee peru enti? Okka line lo cheppu.").content)
```

```python
question = "Ippudu Jaipur lo weather enti? Hyderabad nundi Jaipur ki ee weekend flight price entha?"
print(llm.invoke(question).content)
```

```python
from langchain_core.tools import tool
from langgraph.prebuilt import create_react_agent

@tool
def search_flights(city: str) -> str:
    """Hyderabad nundi given city ki weekend round-trip flight price (INR) return chestundi."""
    prices = {"jaipur": 5200, "goa": 4800, "delhi": 4500}
    return f"Flight to {city}: Rs {prices.get(city.lower(), 9000)} round trip"

@tool
def search_hotels(city: str) -> str:
    """Given city lo 2 nights hotel cost (INR) return chestundi."""
    prices = {"jaipur": 4000, "goa": 6500, "delhi": 5000}
    return f"Hotel in {city}: Rs {prices.get(city.lower(), 8000)} for 2 nights"

@tool
def check_budget(total_cost: int, budget: int) -> str:
    """Total cost budget lopu undo ledo check chestundi."""
    return "Within budget" if total_cost <= budget else f"Over budget by Rs {total_cost - budget}"

agent = create_react_agent(llm, tools=[search_flights, search_hotels, check_budget])
print("Agent ready!")
```

```python
q = "Rs 15000 lopu weekend trip plan cheyyi. Jaipur leda Goa lo edi budget lo fit avthundi?"
result = agent.invoke({"messages": [("user", q)]})
print(result["messages"][-1].content)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", api_version="v1beta")
agent = create_react_agent(llm, tools=[search_flights, search_hotels, check_budget])

q = "Rs 15000 lopu weekend trip plan cheyyi. Jaipur leda Goa lo edi budget lo fit avthundi?"
result = agent.invoke({"messages": [("user", q)]})
print(result["messages"][-1].content)
```

```python
print(result["messages"][-1].text)
```

```python
for step in agent.stream({"messages": [("user", q)]}, stream_mode="values"):
    step["messages"][-1].pretty_print()
```
